```python
print("Task 1: Speed Unit Conversion")

speed_kmh = float(input("Enter speed [km/h]: "))
speed_ms = speed_kmh / 3.6

print(f"Speed [m/s] = {speed_ms:.2f} m/s")
```

```python
print("Task 2: Power Unit Conversion")

Power_kW = float(input("Power[kW]: "))
Power_W = Power_kW * 1000

print(f"Power = {Power_W:.2f} W")
```

```python
print("Task 3: Electrical Power and Energy Calculator")
voltage_V = float(input("Voltage [V]: "))

current_A = float(input("Current [A]: "))

time_hour = float(input("Operating Time [Hour]: "))

power_W = voltage_V * current_A

energy_Wh = power_W * time_hour

print(" ")
print("===== ELECTRICAL CALCULATION =====")
print(f"Voltage : {voltage_V:.2f} V")
print(f"Current : {current_A:.2f} A")
print(f"Power : {power_W:.2f} W")
print(f"Operating Time : {time_hour:.2f} hour")
print(f"Energy : {energy_Wh:.2f} Wh")
```

```python
print("Task 4: Comparison and Logical Operators")
voltage_V = float(input("Voltage [V]: "))

current_A = float(input("Current [A]: "))

time_hour = float(input("Operating Time [Hour]: "))

power_W = voltage_V * current_A

energy_Wh = power_W * time_hour

power_W_limit = 200
voltage_V_limit =24

is_high_power = power_W > power_W_limit
is_high_voltage = voltage_V > voltage_V_limit

print(" ")
print("===== ELECTRICAL CALCULATION TEST =====")
print(f"Voltage : {voltage_V:.2f} V")
print(f"Current : {current_A:.2f} A")
print(f"Power : {power_W:.2f} W")
print(f"Operating Time : {time_hour:.2f} hour")
print(f"Energy : {energy_Wh:.2f} Wh")
print(f"High power : {is_high_power}")
print(f"High voltage : {is_high_voltage}")
print(f"Both conditions : {is_high_power and is_high_voltage}")
print(f"Any condition : {is_high_power or is_high_voltage}")
```

# Short Reflection

1. Apa perbedaan int dan float?

int merupakan tipe data berupa bilangan bulat, sedangkan float merupakan tipe data berupa bilangan bulat atau pecahan.

2. Mengapa hasil input() perlu dikonversi menggunakan float() untuk perhitungan numerik?

Hasil dari input() perlu dikonversi menjadi float() karena input() secara otomatis membaca perintah yang dimasukkan sebagai tipe data str atau string.

3. Apa perbedaan operator = dan == dalam Python?

Perbedaan operator = dan == adalah operator = merupakan suatu operator untuk mendefinisikan suatu nilai ke dalam sebuah variabel, sedangkan operator == merupakan suatu operator untuk membandingkan 2 buah nilai

4. Apa fungsi operator and dan or?

Perbedaan antara kedua operator tersebut adalah operator and hanya akan menghasilkan true jika semua kondisi yang dihubungkan bernilai benar, sedangkan operator or akan menghasilkan true jika salah satu kondisi benar
